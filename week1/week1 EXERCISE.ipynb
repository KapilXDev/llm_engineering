```python
# imports
# If these fail, please check you're running from an 'activated' environment with (llms) in the command prompt

import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from scraper import fetch_website_links, fetch_website_contents
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
openai=OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
system_prompt = """"
You are a phd level teacher in technical department. 
So whenever you get a technical question you respond with a good explaination. 
So that your students not only understand but absorb your answer to that question.
Help them visualize the answer.
"""
```

```python
def generate_user_prompt(ques):
    user_prompt = f"""
    please give an answer and explain it in detail. So that theres no room for doubt.
    Please help to visualize the answer.
    """  
    user_prompt += ques;
    return user_prompt
```

```python
messages = [
    {"role": "system", "content":system_prompt},
    {"role": "user", "content":generate_user_prompt(question)}
]
```

```python
# Get gpt-4o-mini to answer, with streaming
```

```python
# Get Llama 3.2 to answer
stream = openai.chat.completions.create(
    model=MODEL_LLAMA,
    messages=messages,
    stream=True
)

response=""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=display_handle.display_id)

```
